```python
from dotenv import load_dotenv
from pydantic import BaseModel, Field
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.output_parsers import JsonOutputParser
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
from typing import Literal

load_dotenv()

# Simple Output Model
class SupervisorOutput(BaseModel):
    agent_name: Literal["frontend", "backend", "database"] = Field(
        description="The agent to be called next"
    )
    task_details: str = Field(
        description="Detailed explanation of what needs to be coded next"
    )
    status_message: str = Field(
        description="Short progress update (2-3 sentences) including the filename where the next code changes should be applied"
    )

# Simple Supervisor Prompt
supervisor_prompt = ChatPromptTemplate.from_messages([
    (
        "system",
        """You are a software project manager supervising an AI coding team.

Your team: [frontend, backend, database]

You receive:
- Directory structure: {directory}
- Implementation flow: {flow}
- Conversation history between agents which tracks the progress of what work has been done to guide the next task.

Your job:
1. Look at what's been done in the conversation
2. Decide which agent should work next
3. Tell them exactly what to code along with the languages to be used.
4. Give a short status update

Rules:
- Work step by step, not everything at once
- Build backend before frontend that uses it
- Database setup before backend that uses database
- One clear task per agent call

{formatinstructions}
"""
    ),
    MessagesPlaceholder(variable_name="messages"),
])

# Create supervisor chain
llm = ChatGoogleGenerativeAI(model="gemini-2.0-flash-exp")  
supervisor_parser = JsonOutputParser(pydantic_object=SupervisorOutput)
supervisor_chain = supervisor_prompt | llm | supervisor_parser
initial_messages = [
    {"role":"human","content":"We have created the project directory and you can begin with the project"}
]
# Simple test for starting project
result = supervisor_chain.invoke({
    "directory": directory,
    "flow": flow,
    "messages": initial_messages,  # Empty messages for starting
    "formatinstructions": supervisor_parser.get_format_instructions()
})

print(result)
```

```python
from typing import TypedDict, Annotated, List, Union
from langchain_core.messages import HumanMessage, AIMessage, SystemMessage
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
from langchain_core.tools import Tool
from langchain_openai import ChatOpenAI
from langgraph.graph import StateGraph, END
from langgraph.prebuilt import ToolExecutor, ToolInvocation
from langgraph.graph.message import add_messages
import json

# Define the agent state
class AgentState(TypedDict):
    messages: Annotated[List[Union[HumanMessage, AIMessage]], add_messages]
    directory: str
    command: str
    design_config: str
    tech_stack: str
    iteration_count: int
    max_iterations: int

# Tool definitions
def read_file_tool(file_path: str) -> str:
    """Read contents of a file"""
    try:
        with open(file_path, 'r', encoding='utf-8') as f:
            return f.read()
    except Exception as e:
        return f"Error reading file: {str(e)}"

def write_file_tool(file_path: str, content: str) -> str:
    """Write content to a file"""
    try:
        with open(file_path, 'w', encoding='utf-8') as f:
            f.write(content)
        return f"Successfully wrote to {file_path}"
    except Exception as e:
        return f"Error writing file: {str(e)}"

def search_documentation_tool(query: str) -> str:
    """Search for documentation or syntax help"""
    # This would integrate with your search functionality
    return f"Documentation search results for: {query}"

def validate_syntax_tool(code: str, language: str) -> str:
    """Validate code syntax"""
    # This would integrate with syntax validation
    return f"Syntax validation results for {language} code"

# Create tools
tools = [
    Tool(
        name="read_file",
        description="Read the contents of a file",
        func=read_file_tool
    ),
    Tool(
        name="write_file", 
        description="Write content to a file",
        func=write_file_tool
    ),
    Tool(
        name="search_docs",
        description="Search documentation for syntax or library information",
        func=search_documentation_tool
    ),
    Tool(
        name="validate_syntax",
        description="Validate code syntax",
        func=validate_syntax_tool
    )
]

tool_executor = ToolExecutor(tools)

# ReAct prompt template
react_prompt = ChatPromptTemplate.from_messages([
    (
        "system",
        """You are a highly capable **Front-End Developer Agent** using the ReAct (Reasoning + Acting) framework.

Your role:
- Receive coding instructions from the Supervisor Agent
- Use reasoning and actions to implement the requested feature

Given:
- Directory structure: {directory}
- Coding command: {command}
- Design configuration: {design_config}

**ReAct Process:**
1. **Reason**: Think through the problem step by step
2. **Act**: Use tools to gather information or write code
3. **Observe**: Analyze the results of your actions
4. **Repeat**: Continue until the task is complete

**Available Tools:**
- read_file: Read existing files for reference
- write_file: Write code to files
- search_docs: Search for documentation/syntax
- validate_syntax: Validate your code --> internet search

**Output Format:**
Always structure your response as:

**Thought**: [Your reasoning about what to do next]
**Action**: [The tool you want to use]
**Action Input**: [The input to the tool]

After receiving the observation, continue with:
**Observation**: [What you learned from the action]

When complete, end with:
**Final Answer**: [Summary of what was built and file paths modified]

**Guidelines:**
- Follow design config strictly
- Write modular, scalable code
- Be responsive and accessible
- Validate your approach before coding
- Ask for clarification if needed

Start by reasoning about the task and planning your approach.
"""
    ),
    MessagesPlaceholder(variable_name="messages"),
])

# Initialize the LLM
llm = ChatOpenAI(model="gpt-4", temperature=0)

def reasoning_node(state: AgentState):
    """The reasoning node where the agent thinks about what to do next"""
    
    # Check iteration limit
    if state["iteration_count"] >= state["max_iterations"]:
        return {
            "messages": [AIMessage(content="Maximum iterations reached. Task may need manual intervention.")],
            "iteration_count": state["iteration_count"] + 1
        }
    
    # Format the prompt with state information
    formatted_prompt = react_prompt.format_messages(
        directory=state["directory"],
        command=state["command"],
        design_config=state["design_config"],
        tech_stack=state["tech_stack"],
        messages=state["messages"]
    )
    
    # Get the LLM response
    response = llm.invoke(formatted_prompt)
    
    return {
        "messages": [response],
        "iteration_count": state["iteration_count"] + 1
    }

def action_node(state: AgentState):
    """The action node where the agent uses tools"""
    
    # Get the last message from the agent
    last_message = state["messages"][-1]
    
    # Parse the action from the message
    action_text = last_message.content
    
    # Simple parsing - in production, you'd want more robust parsing
    if "**Action**:" in action_text and "**Action Input**:" in action_text:
        try:
            action_start = action_text.find("**Action**:") + len("**Action**:")
            action_end = action_text.find("**Action Input**:")
            action_name = action_text[action_start:action_end].strip()
            
            input_start = action_text.find("**Action Input**:") + len("**Action Input**:")
            action_input = action_text[input_start:].strip()
            
            # Remove any trailing text after the input
            if "**" in action_input:
                action_input = action_input.split("**")[0].strip()
            
            # Execute the tool
            tool_invocation = ToolInvocation(
                tool=action_name,
                tool_input=action_input
            )
            
            result = tool_executor.invoke(tool_invocation)
            
            # Format the observation
            observation_message = AIMessage(
                content=f"**Observation**: {result}"
            )
            
            return {"messages": [observation_message]}
            
        except Exception as e:
            error_message = AIMessage(
                content=f"**Observation**: Error executing action: {str(e)}"
            )
            return {"messages": [error_message]}
    
    # If no valid action found, continue reasoning
    return {"messages": []}

def should_continue(state: AgentState) -> str:
    """Determine whether to continue the ReAct loop or end"""
    
    last_message = state["messages"][-1]
    
    # Check if we have a final answer
    if "**Final Answer**:" in last_message.content:
        return "end"
    
    # Check iteration limit
    if state["iteration_count"] >= state["max_iterations"]:
        return "end"
    
    # Check if we need to take an action
    if "**Action**:" in last_message.content:
        return "action"
    
    # Continue reasoning
    return "reasoning"

# Build the graph
workflow = StateGraph(AgentState)

# Add nodes
workflow.add_node("reasoning", reasoning_node)
workflow.add_node("action", action_node)

# Set entry point
workflow.set_entry_point("reasoning")

# Add edges
workflow.add_conditional_edges(
    "reasoning",
    should_continue,
    {
        "action": "action",
        "end": END,
        "reasoning": "reasoning"
    }
)

workflow.add_conditional_edges(
    "action",
    should_continue,
    {
        "reasoning": "reasoning",
        "end": END
    }
)

# Compile the graph
app = workflow.compile()

# Usage example
def run_frontend_agent(directory: str, command: str, design_config: str, tech_stack: str):
    """Run the frontend ReAct agent"""
    
    initial_state = {
        "messages": [HumanMessage(content=f"Please implement: {command}")],
        "directory": directory,
        "command": command,
        "design_config": design_config,
        "tech_stack": tech_stack,
        "iteration_count": 0,
        "max_iterations": 10
    }
    
    # Execute the workflow
    result = app.invoke(initial_state)
    
    return result

# Example usage
if __name__ == "__main__":
    result = run_frontend_agent(
        directory="/project/frontend",
        command="Create a responsive login form component",
        design_config="Material Design, primary color: #1976d2, font: Roboto",
        tech_stack="React, TypeScript, Material-UI"
    )
    
    print("Final result:", result["messages"][-1].content)
```
